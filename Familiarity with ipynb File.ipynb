This notebook demonstrates basic usage of Jupyter Notebook, including importing a dataset using pandas and visualizing data with matplotlib.


```python
#Import pandas and matplotlib

import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Load a CSV file 

# Example data
data = {
    'Category': ['AI', 'Data Science', 'Machine learning', 'Gen AI', 'Data Science', 'Machine learning', 'AI'],
}
df = pd.DataFrame(data)
df

```

```python
# plot a graph

import matplotlib.pyplot as plt

category_counts = df['Category'].value_counts()

category_counts.plot(kind='bar')

plt.title('Article Categories')

plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```
